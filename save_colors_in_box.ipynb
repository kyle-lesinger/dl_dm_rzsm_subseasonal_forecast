```python
%matplotlib inline
import matplotlib.pyplot as plt
import matplotlib.patches as patches

# Define the named colors
colors = {
    'DL-DM': 'blue',
    'DL-BC_DM': 'green',
    'DL': 'purple'
}

# Create a figure and axis
fig, ax = plt.subplots(figsize=(10, 2))
ax.set_xlim(0, 3)
ax.set_ylim(0, 1)
ax.axis('off')  # Hide the axes

# Draw rectangles
for i, (name, color) in enumerate(colors.items()):
    rect = patches.Rectangle((i, 0), 1, 1, facecolor=color)
    ax.add_patch(rect)
    ax.text(i + 0.5, 0.5, name, va='center', ha='center', fontsize=40, color='white', weight='bold')

plt.tight_layout()
plt.savefig("colors_horizontal.png", dpi=300)
plt.show()
```

```python
%matplotlib inline
import matplotlib.pyplot as plt
import matplotlib.patches as patches


# Define the named colors and their RGB values (matplotlib defaults)
colors = {
    'DL-DM': 'blue',
    'DL-BC_DM': 'green',
    'DL': 'purple'
}


# Create a vertical layout
fig, ax = plt.subplots(figsize=(10, 6))
ax.set_xlim(0, 1)
ax.set_ylim(0, len(colors))
ax.axis('off')  # Hide axes

# Draw vertically stacked rectangles
for i, (name, color) in enumerate(colors.items()):
    rect = patches.Rectangle((0, len(colors) - i - 1), 1, 1, facecolor=color)
    ax.add_patch(rect)
    ax.text(0.5, len(colors) - i - 0.5, name, va='center', ha='center', fontsize=20, color='white', weight='bold')

plt.tight_layout()
plt.savefig("colors_vertical.png", dpi=300)
plt.show()
```
